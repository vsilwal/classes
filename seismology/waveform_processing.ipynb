### What is a Stream?

ObsPy reads waveform data into a **Stream** object.

A Stream is a collection of one or more **Trace** objects.

A Trace contains:

- the seismic waveform data
- station information
- channel information
- start and end time
- sampling rate
- number of samples

We can inspect the first Trace using `st[0]`.


````python
print(st[0])
print()
print(st[0].stats)
````

````python
tr = st[0]

print("Trace ID      :", tr.id)
print("Start time    :", tr.stats.starttime)
print("End time      :", tr.stats.endtime)
print("Sampling rate :", tr.stats.sampling_rate, "Hz")
print("Number samples:", tr.stats.npts)
````

# Part 3 — Removing the Mean


````python
from obspy import read
data_raw1 = read('data/PA01.bhy')  ## Read waveform files into an ObsPy Stream object.

# Plot the waveform without any processing and copy the stream.
data_raw1.plot()
data_processed1 = data_raw1.copy()  ## Copy the raw stream to the new one to be processed
````

Now let’s detrend the waveforms and plot it again.

````python
data_processed1.detrend("demean")
data_processed1.plot()

# You can compare the raw and processed waveforms by overlapping them.
import matplotlib.pyplot as plt

plt.figure(figsize=(13,5))
for tr in data_raw1:
    data1 = tr.data
plt.plot(data1,color='red',label='raw')

for tr in data_processed1:
    data2 = tr.data
plt.plot(data2,color='blue',label='removed mean')

plt.legend()
````

# Part 4 — Removing a Linear Trend

Sometimes a waveform contains a gradual increase or decrease with time. This is called a **linear trend**.

We can remove it using:

```python
detrend("linear")
```


````python
data_raw2 = read('data/LLT.E.Vel.BF.SAC')
data_raw2.plot()
data_processed2 = data_raw2.copy()
````

````python
data_processed2.detrend("linear")
data_processed2.plot()

# You can compare the raw and processed waveforms by overlapping them.
import matplotlib.pyplot as plt

plt.figure(figsize=(13,5))
for tr in data_raw2:
    data1 = tr.data
plt.plot(data1,color='red',label='raw')

for tr in data_processed2:
    data2 = tr.data
plt.plot(data2,color='blue',label='linear detrended')

plt.legend()
````

### Demean vs linear detrend

- **Demean:** removes a constant offset.
- **Linear detrend:** removes a gradual linear trend.

# Part 5 — Filtering Seismic Data

A seismic waveform contains energy at many frequencies.

Filtering allows us to emphasize the frequency range of interest.

Common filters include:

- **Low-pass:** keeps lower frequencies and attenuates higher frequencies.
- **High-pass:** keeps higher frequencies and attenuates lower frequencies.
- **Band-pass:** keeps a selected frequency band.

We will first demonstrate a low-pass filter.


````python
st[0].plot()

processed = st[0].copy()

# Apply a low-pass filter or try different filters
processed.filter("lowpass", freq=1)

# Plot the filtered waveform
processed.plot()
````

````python
# Plot the filtered waveform for the selected time window
# Get the starting time
start_time = processed.stats.starttime
processed.plot(
    starttime=start_time + 300,
    endtime=start_time + 1600
)
````
